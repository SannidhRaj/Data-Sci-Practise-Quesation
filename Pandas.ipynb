Q11. Filter Students

Using the following DataFrame:

| Name  | Age | Marks |
|:------|----:|------:|
| Rahul | 20  | 85    |
| Aman  | 21  | 78    |
| Priya | 19  | 92    |
| Neha  | 20  | 88    |
| Rohan | 22  | 65    |

**Task:**  
Find all students who scored **more than 80 marks**.

```python
data={
    "Name":["Rahul","Aman","Priya","Neha","Rohan"],
    "Age":[20,21,19,20,22],
    "Marks":[95,78,92,88,65]
    
}
df=pd.DataFrame(data)
df
```

```python
print( df[df["Marks"]>80])
```

Q12. Multiple Conditions<br>
Find students who:<br>
scored more than 80,
AND are younger than 21

```python
print("scored more than 80 AND are younger than 21")
print(df[(df["Marks"]>80) & (df["Age"]<21)])
```

Q13. Filter Using OR<br>
Find students who are:<br>
younger than 20,
OR scored more than 90.

```python
print("students who are younger than 20 OR scored more than 90.")
print(df[(df["Age"]<20) | (df["Marks"]>90)])

#METHOD 2(QUERY METHOD)
print(df.query("Marks>90 | Age<20")) 
```

Q14. Find Maximum Marks<br>
Find:<br>
Highest marks,
Student who obtained the highest marks,
Row containing the highest marks.

```python
print("Highest Marks",df["Marks"].max())
print("student who securec highest marks:",df.loc[df["Marks"].idxmax(),"Name"],"\n")

print("row containg highest marks")
print(df.loc[[df["Marks"].idxmax()]])
```

Q15. Find Minimum Marks<br>
Find:<br>
Lowest marks,
Student with the lowest marks,
Complete information of that student

```python
print("lowest marks",df["Marks"].min())
print("student with lowest marks:",df.loc[df["Marks"].idxmin() ,"Name"])
print("complet info""\n",df.loc[df["Marks"].idxmin()])#to get output in table like above use extra square bracket it convert to data frame
```

Q16. Sorting
Sort the DataFrame:<br>
By marks in ascending order,
By marks in descending order,
By age and then marks.

```python
df2=df.copy()#we use copy 
print("Marks in Ascending")
print(df2.sort_values("Marks"))

print("\n""Marks in Descending")
print(df2.sort_values("Marks", ascending=False),"\n")

print("by Age then Marks ")
print(df.sort_values(["Age","Marks"]))
```

Q17. Add a New Column<br>
Add a column called Passed.<br>
Condition:<br>
Marks >= 40 → Pass<br>
Marks < 40 → Fail

```python
df2["Passed"]=df2["Marks"].apply(lambda x:"Pass" if x >=40 else "Fail")
df2
```

Q18. Create Grade Column<br>
Create a Grade column using:<br>
90+       → A<br>
80–89     → B<br>
70–79     → C<br>
60–69     → D<br>
Below 60  → F<br>

```python
df3=df2.copy()
df3["Grade"]=df3["Marks"].apply(
    lambda x: 
        "A" if x>90 else
        "B" if 80 <= x <= 89 else
        "C" if 70 <= x <=79  else
        "D" if 60 <= x <=69  else
        "F"
        )
df3
```
