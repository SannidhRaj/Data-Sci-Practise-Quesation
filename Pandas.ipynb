Q19. Missing Values <br>
Create this DataFrame:<br>
data = {<br>
    "Name": ["Rahul", "Aman", "Priya", "Neha", "Rohan"],<br>
    "Age": [20, np.nan, 19, 20, np.nan],<br>
    "Marks": [85, 78, np.nan, 88, 65]<br>
}<br>
Find:<br>
Number of missing values in each column,
Total number of missing values,
Rows containing missing values.

```python
import numpy as np
data = {
    "Name": ["Rahul", "Aman", "Priya", "Neha", "Rohan"],
    "Age": [20, np.nan, 19, 20, np.nan],
    "Marks": [85, 78, np.nan, 88, 65]
}
df=pd.DataFrame(data)
df
```

```python
print("totaL No of missing value in each column")
print(df.isnull().sum(),"\n")

print("Total missing value",df.isnull().sum().sum())

print("Rows containg missing value")
print(df[df.isnull().any(axis=1)])
```

Q20. Handle Missing Values
Using the same DataFrame:<br>
Replace missing Age with the average age,
Replace missing Marks with the average marks.

```python
df2=df.copy()
avg_age=int(df2["Age"].mean())
df2["Age"]=df2["Age"].fillna(avg_age)
print(df2["Age"].fillna(avg_age))

avg_marks=df2["Marks"].mean()
df2["Marks"]=df2["Marks"].fillna(avg_marks)
print(df2["Marks"].fillna(avg_marks))
```

Q29. Basic Statistics
Given:
df["Marks"] = [45, 60, 75, 90, 30, 85, 70]<br>
Find:<br>
Mean,
Median,
Mode,
Standard deviation,
Variance,
Minimum,
Maximum.

```python
df=pd.DataFrame({"Marks" : [45, 60,60, 75, 90, 30, 85, 70]})

mean=df["Marks"].mean()
median=df["Marks"].median()
mode=df["Marks"].mode().tolist()
std_dev=df["Marks"].std()
variance=df["Marks"].var()
min=df["Marks"].min()
max=df["Marks"].max()

print("Mean",mean )
print("Median",median )
print("Mode ",mode )
print("Standard deviation ",std_dev)
print("Variance",variance)
print("Minimum ",min )
print("Maximum",max  )
```
