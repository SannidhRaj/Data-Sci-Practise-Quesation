Q3. Custom Index<br>

Q4. Access Series Values
Using the previous Series:
Access the marks of Science.
Access the marks of Computer.
Access the first two values.

```python
ser=pd.Series([85,90,78,92],index=['maths','sci','eng','comp'])
print("marks of sci",ser["sci"])
print("marks of COMP",ser["comp"])
print("marks of first two",ser[["maths","sci"]])
```

Q6. DataFrame Properties<br>
Using the DataFrame from Q5, find:<br>
Number of rows,
Number of columns,
Shape,
Column names,
Index,
Data types.

```python
data={
    "Name":["Rahul","Aman","priya","neha"],
    "Age":[20,21,19,20],
    "Marks":[85,78,92,88]

}
df=pd.DataFrame(data)
print("no of row",df.shape[0])
print("no. of column",df.shape[1])
print("shape",df.shape)
print("column names",df.columns.tolist())
print("index ",df.index)
print("data type",df.dtypes)
```

Q7. View Data<br>
Using the same DataFrame, display:<br>
First 2 rows,
Last 2 rows,
First 3 rows,
Use appropriate Pandas methods.

```python
data={
     "Name":["Rahul","Aman","priya","neha"],
    "Age":[20,21,19,20],
    "Marks":[85,78,92,88]
}
df=pd.DataFrame(data)
print("First 2 rows =""\n", df.iloc[:2],"\n")

print("last 2 rows" "\n",df.iloc[-2:],"\n")
print("first 3 rows""\n",df.head(3))
```

Q8. Select Columns<br>
From the DataFrame, select:<br>
Only Name,
Only Marks,
Name and Marks,

```python
data={
     "Name":["Rahul","Aman","priya","neha"],
    "Age":[20,21,19,20],
    "Marks":[85,78,92,88]
}
df=pd.DataFrame(data)
print("only name =",df['Name'].tolist())
print("only marks =",df['Marks'].tolist())
print("only name and Marks ""\n",df[['Name','Marks']])
```

Q9. Select Rows<br>

```python
print("first row""\n",df.iloc[:1])
print("last Row""\n",df.iloc[-1:])
print("first 3 rows""\n",df.iloc[:3])
```

Q10. Select Using Labels<br>
Set Name as the index.
Then use .loc to access:<br>
Rahul's complete information,
Priya's marks,
Rahul and Neha's information.

```python
df2=df.set_index("Name")
print("Rahul complete info""\n",df2.loc["Rahul"],"\n")
print("priya marks",df2.loc["priya","Marks"],"\n")
print("Rahul and Neha complete info""\n",df2.loc[["Rahul","neha"]],"\n")
```
