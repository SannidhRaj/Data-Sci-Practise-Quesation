Q11 Reverse an Array
Given:
arr = np.array([10, 20, 30, 40, 50])
Reverse the array using NumPy slicing.<BR>
Expected:
[50 40 30 20 10]

```python
arr=np.arange(10,51,10) #STAER,STOP,STEP
print(arr)
rev_arr=arr[::-1] #SLICING FORMAT(START,STOP,STEP)
print(rev_arr)
```

Q12 Boolean Masking<BR>
Given:<BR>
arr = np.array([12, 45, 7, 89, 23, 56, 3, 78])<BR>
Extract all values greater than 50.<br>
Using the same array, extract values that are:
greater than 20 AND less than 80.

```python
arr=np.array([12,45,89,23,56,3,78])
print(arr)
arr2=[]
for i in arr:
    if i >50:
        arr2.append(i)
print(arr2)    

for i in arr:
    if 20<i<80:
        arr2.append(i)
print(arr2)    
    
```
