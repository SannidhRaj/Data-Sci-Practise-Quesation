| ### Standardization<br><br>Uses mean and standard deviation.<br><br>Formula:<br>`python<br>(x - mean) / std<br>`<br><br>`z = 0` → value equals mean<br>`z > 0` → above mean<br>`z < 0` → below mean                        | ### Min-Max Normalization<br><br>Uses minimum and maximum values.<br><br>Formula:<br>`python<br>(x - min) / (max - min)<br>`<br><br>Usually transforms values to the range `0–1`.                                                                |
| ### Standardization vs Normalization<br><br>**Standardization** → Mean + Standard Deviation<br><br>**Min-Max** → Minimum + Maximum                                                                                         | ### Counting Conditions<br><br>`True → 1`<br>`False → 0`<br><br>Therefore:<br>`python<br>np.sum(data > 30)<br>`<br>can count how many values satisfy the condition.                                                                              |

<br>

Each row represents a student and each column represents a subject.

<b>Find:</b>

1. Average marks of each student
2. Average marks of each subject
3. Highest mark
4. Student with the highest average
5. Students whose average is greater than 75

````python
import numpy as np

marks = np.array([
    [78, 85, 90],
    [65, 70, 72],
    [92, 88, 95],
    [55, 60, 58]
])

# 1. Average marks of each student
student_avg = np.mean(marks, axis=1)
print("Average marks of each student:", student_avg)

# 2. Average marks of each subject
subject_avg = np.mean(marks, axis=0)
print("Average marks of each subject:", subject_avg)

# 3. Highest mark
highest_mark = np.max(marks)
print("Highest mark:", highest_mark)

# 4. Student with the highest average
highest_avg = np.max(student_avg)
highest_avg_student = np.argmax(student_avg) + 1
print("Highest average:", highest_avg)
print("Student with the highest average: Student", highest_avg_student)

# 5. Students whose average is greater than 75
students_above_75 = np.where(student_avg > 75)[0] + 1
print("Students whose average is greater than 75:", students_above_75)
````

Q22 Temperature Data
Given:<BR>
temperature = np.array([32, 35, 31, 29, 40, 38, 36])<BR>
Find:<BR>
Average temperature<BR>
Maximum temperature<BR>
Minimum temperature<BR>
Days where temperature was above 35<BR>
Number of days above 35<BR>

````python
temperature = np.array([32, 35, 31, 29, 40, 38, 36])

avg_temp=np.mean(temperature)
min_temp=np.min(temperature)
max_temp=np.max(temperature)
temp_above_35=temperature[temperature>35]
day_above_35=np.sum(temperature>35)

print("AVG TEMP=",avg_temp)
print("MIN TEMP=",min_temp)
print("MAX TEMP=",max_temp)
print("TEMP ABOVE 35=",temp_above_35)
print("Number of days above 35:", day_above_35)
````

Q23. Sales Data


Analyze the following monthly sales data using NumPy.


Given:

| Month | Product 1 | Product 2 | Product 3 |
|:-----:|----------:|----------:|----------:|
| Month 1 | 100 | 200 | 150 |
| Month 2 | 300 | 250 | 400 |
| Month 3 | 200 | 180 | 220 |

 Find:

1. **Total sales**
2. **Total sales for each month**
3. **Total sales for each product**
4. **Month with the highest sales**
5. **Product with the highest total sales**

**Note:**
- Rows represent **months**.
- Columns represent **products**.
- Use NumPy functions such as `np.sum()`, `np.argmax()`, and array operations wherever appropriate.

````python
sales = np.array([
    [100, 200, 150],
    [300, 250, 400],
    [200, 180, 220]
])
total_sales=np.sum(sales)
total_sale_each_month=np.sum(sales,axis=1)
total_sale_each_product=np.sum(sales,axis=0)

month_sale=np.sum(sales,axis=1)
month_wise_highest_sale=np.argmax(month_sale)+1

product_sale=np.sum(sales,axis=0)
product_wise_highest_sale=np.argmax(product_sale)+1




print("Total sales =",total_sales)
print("Total sales for each month =",total_sale_each_month)
print("Total sales for each product =",total_sale_each_product)
print("Month with the highest sales = Month",month_wise_highest_sale)
print("Product with the highest total sales = product",product_wise_highest_sale)
````

Given:<BR>
data = np.array([10, 20, 30, 40, 50])<BR>
Calculate the standardized values using:<BR>
z = (x - mean) / standard_deviation<BR>
Do this using NumPy functions.
This is directly connected to Machine Learning preprocessing.

````python
data = np.array([10, 20, 30, 40, 50])

mean=np.mean(data)
std_dev=np.std(data)

z=(data-mean)/std_dev

print("MEAN =",mean)
print("STANDARD DEVIATION =",std_dev)
print("STANDARDIZED VALUE =",z)
````

| 15 | 25 | 35 |
| 20 | 30 | 40 |
| 25 | 35 | 45 |
| 30 | 40 | 50 |

 📝 Dataset Representation

```python
data = np.array([
    [10, 20, 30],
    [15, 25, 35],
    [20, 30, 40],
    [25, 35, 45],
    [30, 40, 50]
])
Treat each row as a data sample and each column as a feature.
Find:
Number of samples
Number of features
Mean of each feature
Maximum of each feature
Minimum of each feature
Standard deviation of each feature
Samples where the second feature > 30
Normalize each feature using min-max normalization
normalized = (x - min) / (max - min)

````python
data = np.array([
    [10, 20, 30],
    [15, 25, 35],
    [20, 30, 40],
    [25, 35, 45],
    [30, 40, 50]
])

print("Number of samples",data.shape[0])
print("Number of feature",data.shape[1])
print("Mean of each feature =", np.mean(data, axis=0))
print("Minimum of  each feature",np.min(data ,axis=0))
print("Maxium of  each feature",np.max(data ,axis=0))
print("Standard Devation of each feature",np.std(data ,axis=0))
print("Samples where second feature > 30 =" ,data[data[:, 1] > 30])

min=np.min(data,axis=0)
max=np.max(data ,axis=0)
nor=(data-min)/(max-min)

print("Normalized value", nor)
````
